# Predictive Customer Churn Classifier – Random Forest & XGBoost

## 1. Project Overview

This project was completed as part of the **Datheon Machine Learning & Data Science Internship – Task 2**.

The project focuses on developing a predictive customer churn classification system using two ensemble machine learning algorithms: **Random Forest and XGBoost**.

Customer churn occurs when customers discontinue a service or subscription. For telecommunications businesses, identifying customers who are likely to leave is important for improving customer retention, reducing customer acquisition costs and supporting business decision-making.

The objective was to train and evaluate machine learning models that can predict customer churn using historical telecommunications subscription records.

**Project achievement:** The XGBoost classifier achieved a test ROC-AUC score of **84.17%**, exceeding the project's required benchmark of 80%.

## 2. Business Problem

Telecommunications companies face the ongoing challenge of retaining customers in a competitive market.

The business problem addressed in this project is:

*How can a telecommunications company identify customers who are likely to discontinue their subscriptions so that appropriate customer retention strategies can be implemented?*

This is a supervised binary classification problem.

The target variable is:
- `Yes` – Customer has churned.
- `No` – Customer has remained with the company.

The predictive models learn patterns from historical customer information and use those patterns to estimate the likelihood of churn.

## 3. Project Objectives

The main objectives were to:
1. Load and inspect the Telco Customer Churn dataset.
2. Clean the dataset and convert `TotalCharges` into a numeric format.
3. Perform exploratory data analysis (EDA).
4. Preprocess categorical variables using OneHotEncoder.
5. Address class imbalance using class weighting.
6. Split the dataset into 80% training and 20% testing.
7. Train Random Forest and XGBoost classifiers.
8. Perform five-fold stratified cross-validation.
9. Evaluate model performance using ROC-AUC, accuracy, precision, recall and F1-score.
10. Visualise ROC curves and confusion matrices.
11. Identify the ten most influential features for each model.
12. Select the best-performing model based on test ROC-AUC.

## 4. Dataset Description

**Dataset:** Telco Customer Churn

The dataset contains telecommunications customer subscription records, including customer demographics, services, billing information and churn status.

| Dataset information | Details |
|---|---:|
| Total records | 7,043 |
| Original columns | 21 |
| Target variable | Churn |
| Classification type | Binary |
| Non-churn customers | 5,174 |
| Churn customers | 1,869 |

### Selected features

The dataset includes:
- Customer demographics: Gender and Senior Citizen status.
- Customer relationships: Partner and Dependents.
- Subscription information: Tenure and Contract.
- Telecommunications services: Internet Service, Phone Service and Multiple Lines.
- Additional services: Online Security, Online Backup and Tech Support.
- Billing information: Monthly Charges and Total Charges.
- Target: Churn.

The `customerID` column was excluded from model training because it is an identifier rather than a meaningful behavioural predictor.

The `TotalCharges` column was converted to numeric format during data preparation.

### Class distribution

The target variable was imbalanced:

| Customer status | Records | Percentage |
|---|---:|---:|
| Stayed (No) | 5,174 | 73.46% |
| Churned (Yes) | 1,869 | 26.54% |

Class weighting was used to account for this imbalance.

## 6. Methodology

The project followed an end-to-end machine learning workflow.

### Phase 1: Problem Definition and Data Collection
- Defined the customer churn prediction problem.
- Loaded the telecommunications dataset.
- Inspected the dataset structure, columns and data types.

### Phase 2: Data Preparation

**Data cleaning**
- Checked for duplicate records.
- Converted `TotalCharges` into numeric format.
- Excluded the customer identifier from the predictors.
- Prepared numeric and categorical variables.

**Exploratory Data Analysis**

Three visualisations were generated:
1. Customer churn distribution.
2. Churn percentage by contract type.
3. Monthly charges by churn status.

**Feature preprocessing**
- Applied median imputation to numeric features.
- Applied most-frequent imputation to categorical features.
- Used OneHotEncoder to convert categorical variables into numerical features.

**Train/test split**

The dataset was divided using an 80/20 stratified split:
- Training dataset: 5,634 records.
- Testing dataset: 1,409 records.

### Phase 3: Model Development and Evaluation

Two ensemble classification algorithms were implemented.

**Random Forest Classifier**
- 300 decision trees.
- Balanced class weighting.
- Minimum samples per leaf: 2.
- Random state: 42.

**XGBoost Classifier**
- 300 boosting estimators.
- Maximum tree depth: 4.
- Learning rate: 0.05.
- Class imbalance addressed using `scale_pos_weight`.
- Random state: 42.

Both models were evaluated using five-fold stratified cross-validation on the training data. The final models were then evaluated against the independent test dataset.

### Phase 4: Deployment and Monitoring

The best-performing fitted pipeline was saved using Joblib as:

`datheon_customer_churn_model.joblib`

The model could be integrated into a customer-retention dashboard or internal business application.

Future monitoring should include model performance, changes in customer behaviour and data drift.

## 7. Model Performance and Results

### Five-fold cross-validation

| Model | Mean ROC-AUC | Standard deviation |
|---|---:|---:|
| Random Forest | 83.70% | 0.87 percentage points |
| XGBoost | 84.51% | 0.99 percentage points |

XGBoost achieved the higher average cross-validation ROC-AUC.

### Final test-set evaluation

| Evaluation metric | Random Forest | XGBoost |
|---|---:|---:|
| Accuracy | 78.42% | 75.37% |
| Precision | 58.41% | 52.42% |
| Recall | 64.97% | 78.07% |
| F1-score | 61.52% | 62.73% |
| ROC-AUC | 83.48% | **84.17%** |

### Key achievement

**XGBoost achieved a test ROC-AUC of 84.17%, exceeding the required benchmark of 80%.**

This indicates that the model demonstrates good ability to distinguish between customers who churn and customers who remain.

Although Random Forest achieved higher accuracy and precision, XGBoost achieved higher recall. This means XGBoost identified a larger proportion of actual churn cases at the default classification threshold.

For a customer retention use case, higher recall can be valuable because identifying more customers who may leave gives the business more opportunities to intervene.
